```python
def generate_fibonacci(n):
    """n numbers tak ki Fibonacci series generate karta hai"""
    if n <= 0:
        return []
    elif n == 1:
        return [0]

    # Starting ke do numbers
    fib_series = [0, 1]

    # Baaki numbers ke liye loop
    for i in range(2, n):
        next_number = fib_series[-1] + fib_series[-2]
        fib_series.append(next_number)

    return fib_series

# Example usage:
if __name__ == "__main__":
    n_terms = int(input("Kitne terms ki Fibonacci series chahiye? "))
    result = generate_fibonacci(n_terms)
    print(f"Fibonacci Series ({n_terms} terms): {result}")
```
