```python
# User se values lena (Getting values from user in Python)
first_term = int(input("Enter first term (a): "))
common_difference = int(input("Enter common difference (d): "))
n = int(input("Enter number of terms (n): "))

# AP series print karna
print("\nArithmetic Progression (AP) Series:")
for i in range(n):
    term = first_term + (i * common_difference)
    print(term, end=" ")
print() # New line
```
